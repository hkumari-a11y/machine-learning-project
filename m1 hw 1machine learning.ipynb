```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("home_data.csv")
df.head()
```

## Question 1: Selection and Summary Statistics

### 1.1 Zip code with the highest average house sale price

```python
zip_avg = df.groupby("zipcode")["price"].mean()

highest_zip = zip_avg.idxmax()
highest_avg_price = zip_avg.max()

print("Zip code with highest average price:", highest_zip)
print("Highest average price:", highest_avg_price)
```

### 1.2 Average price for houses in that zip code

```python
selected_zip = df[df["zipcode"] == highest_zip]

avg_price_selected_zip = selected_zip["price"].mean()

print("Average price for selected zip code:", avg_price_selected_zip)
```

## Question 2: Filtering Data

### 2.1 Houses with sqft_living greater than 2000 and less than or equal to 4000

```python
filtered_houses = df[(df["sqft_living"] > 2000) & (df["sqft_living"] <= 4000)]

filtered_houses.head()
```

### 2.2 Fraction of all houses in this sqft_living range

```python
fraction = len(filtered_houses) / len(df)

print("Fraction of houses:", fraction)
```

## Question 3: Visualization

### 3.1 Relationship between house sale price and sqft_living

```python
plt.figure(figsize=(8,5))
plt.scatter(df["sqft_living"], df["price"])
plt.xlabel("Sqft Living")
plt.ylabel("House Sale Price")
plt.title("House Sale Price vs Sqft Living")
plt.show()
```

## Conclusion

This section analyzed the King County housing dataset by identifying the zip code with the highest average house sale price and calculating the average price for that area. Houses with living space between 2,000 and 4,000 square feet were filtered, and the fraction of such houses was determined. A scatter plot was also created to visualize the relationship between house sale price and living area.
